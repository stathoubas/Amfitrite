```python
import pandas as pd
import os
import ast
import json
from IPython.display import Image, display, clear_output
```

```python
excel_path = r"C:\Users\KostasPikounis\OneDrive_Inlecom_Personal\OneDrive - INLECOM\Amfitrite\task2\IWD\dataset_summary.xlsx"
# Update the root path to wherever your images are stored. Make sure it ends with a slash!
root = r"C:\Users\KostasPikounis\OneDrive_Inlecom_Personal\OneDrive - INLECOM\Amfitrite\task2\IWD\data"
```

```python
# Load the dataframe
df_o = pd.read_excel(excel_path)

for it, row in df_o.iterrows():
    # Safely convert strings to lists
    try:
        li = ast.literal_eval(str(row["abun"]))
    except (ValueError, SyntaxError):
        continue # Skip if the cell is empty or can't be parsed
    
    # Check your conditions
    if isinstance(li, list) and len(li) > 1 and all([10 < i < 100 for i in li]):
        
        # Clear the previous image and text 
        clear_output(wait=True)
        
        # Print the UID and Class
        print(f"UID: {row.uid}, Class : {row.HAB_status}")
        
        # ---------------------------------------------------------
        # NEW: Read and print compatibility_analysis from metadata
        # ---------------------------------------------------------
        json_path = os.path.join(root, row.uid, "metadata.json")
        
        if os.path.exists(json_path):
            with open(json_path, 'r') as file:
                try:
                    metadata = json.load(file)
                    # Check if the key exists before trying to print it
                    if "compatibility_analysis" in metadata:
                        print("\n--- Compatibility Analysis ---")
                        # json.dumps with indent=4 makes it highly readable
                        print(json.dumps(metadata["compatibility_analysis"], indent=4))
                        print("------------------------------\n")
                    else:
                        print("\n'compatibility_analysis' not found in this metadata.json.\n")
                except json.JSONDecodeError:
                    print("\nError: metadata.json is corrupted or not a valid JSON.\n")
        else:
            print(f"\nMetadata file not found at: {json_path}\n")

        # ---------------------------------------------------------
        # Display the Image
        # ---------------------------------------------------------
        img_path = os.path.join(root, row.uid, "cyfi_prediction_map.png")
        
        if os.path.exists(img_path):
            display(Image(filename=img_path))
        else:
            print(f"Image not found at path: {img_path}")
            
        # Pause and ask the user to proceed
        user_input = input("Press Enter to continue to the next case, or type 'no' to finish: ").strip().lower()
        
        # Break the loop if the user types 'no'
        if user_input == "no":
            print("Iteration finished by user.")
            break
```
