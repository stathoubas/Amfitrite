```python
import pandas as pd
import folium


df = pd.read_csv(r"C:\Users\KostasPikounis\OneDrive_Inlecom_Personal\OneDrive - INLECOM\Amfitrite\task2\OWD\OBIS datasets\2 Harmful Algal Blooms\homogenized_occurrence - Copy.csv")

map_center = [df['decimalLatitude'].mean(), df['decimalLongitude'].mean()]
mymap = folium.Map(location=map_center, zoom_start=12)

for index, row in df.iterrows():
    folium.Marker(
        location=[row['decimalLatitude'], row['decimalLongitude']],
        icon=folium.Icon(color='blue', icon='info-sign')
    ).add_to(mymap)

# 4. Αποθήκευση του χάρτη σε αρχείο HTML
mymap

```
